## Dot product

```python
np.random.seed(10)

mat1 = np.random.randint(10, size=(4,3))
mat2 = np.random.randint(10, size=(4,3))

mat1
```

```python
mat2
```

```python
# Element-wise Multiplication (Hadamard Product)

mat1*mat2
```

```python
# Dot Product

np.dot(mat1, mat2)

# This code will produce error because the dot product works completely diff from the hadamard product. For Dot product to work the inner shape or dimension of the array must be the same, so we might want to flip axeses for that to work
```

```python
#Either reshape or transpose works. so we transpose mat2 to match mat1

mat2.T
```

```python
# Dot Product

mat3 = np.dot(mat1, mat2.T)
mat3
```

```python
mat3.shape
```

## Dot product example (Nut-butter-sales)

```python
np.random.seed(10)

sales_amount = np.random.randint(20, size = (5,3))
sales_amount
```

```python

import pandas as pd
weekly_sales = pd.DataFrame (sales_amount, 
                            index = ["Mon", "Tues", "wed", "Thurs", "Fri"],
                            columns = ["Almond butter", "Peanut butter", "Cashew butter"]
                            )
weekly_sales
```

```python
# Creating array for the price

prices = np.array([10,8,12])
prices
```

```python
prices.shape
```

```python
# Creating prices dataframe

butter_prices = pd.DataFrame(prices.reshape(1,3),
                 index=["Price"],
                 columns = ["Almond butter", "Peanut butter", "Cashew butter"]
                 )
butter_prices
```

```python
total_sales = np.dot(prices,sales_amount.T)
total_sales
```

```python
weekly_sales
```

```python
weekly_sales ["Total($)"] = total_sales
weekly_sales
```

```python
type(total_sales)
```
