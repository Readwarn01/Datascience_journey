```python
import pandas as pd
```

Data structure or objects pandas use to store and manipulate data

- Series >
    A single column of data with labels (called index)
  
- DataFrames >
    A 2-Dimensional table of data made up of rows and column

```python
#Series - 1 Dimensiional
Brand_Colours = pd.Series(["Yellow and Red", "Balck and Green", "Yellow", "Blue"])
Brand_Colours
```

```python
#Series - 1 Dimensiional
Company = pd.Series(["Pwc", "Deloitte", "EY", "KPMG"])
Company
```

```python
#DataFrames - 2 Dimensiional

Big4_Data = pd.DataFrame ({"Company Name": Company, "Brand Colour": Brand_Colours})
Big4_Data
```
